```python
# x = independent variable
# y = dependent variable
# m = slope of the Line
# m = (y2 - y1) / (x2 - x1)
# Slope definition:The ratio of the vertical change to the horizontal change between two points on the line.
import numpy as np
import matplotlib.pyplot as plt
x = np.array([88, 92, 95, 98, 100])
y = np.array([78, 85, 90, 95, 112])
m = np.sum(x - np.mean(x)) * np.sum(y - np.mean(y))
c = np.mean(y) - m * np.mean(x)
print(f"Equation of line: y = {m}x + {c}")
y_pred = m * x + c
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, y_pred, color='red', label='Fitted Line')
plt.xlabel("X")
plt.ylabel("Y")
plt.title('Linear Regression')
plt.legend()
plt.show()
```
